```python
import ee
import geemap

ee.Authenticate()
ee.Initialize(project="sentinel-exploration")
```

```python
from datetime import datetime, timezone

dhaka = ee.Geometry.Rectangle([90.33, 23.68, 90.55, 23.90])

# quick test Sentinel 2 images of Dhaka on septembr 2026
images = (ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
                        .filterBounds(dhaka)
                        .filterDate("2026-09-01", "2026-10-01"))

timestamps = images.aggregate_array("system:time_start").getInfo()

print("SENTINEL 2")
for timestamp in timestamps:
    date = datetime.fromtimestamp(timestamp / 1000, tz=timezone.utc)
    print(date.strftime("%Y-%m-%d"))



# quick test Sentinel 1 images of Dhaka on september 2026
images = (ee.ImageCollection("COPERNICUS/S1_GRD")
                        .filterBounds(dhaka)
                        .filterDate("2026-09-01", "2026-10-01"))

timestamps = images.aggregate_array("system:time_start").getInfo()

print("\nSENTINEL 1")
for timestamp in timestamps:
    date = datetime.fromtimestamp(timestamp / 1000, tz=timezone.utc)
    print(date.strftime("%Y-%m-%d"))
```
